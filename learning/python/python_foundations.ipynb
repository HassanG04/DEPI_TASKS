### Simple Library Management System
Functions to implement:
- show_books() : Displays all available books in the library.
- borrow_book() : Allows the user to borrow a book.
- return_book() : Allows the user to return a previously borrowed book.
- show_borrowed_books() : Displays all books currently borrowed by the user.
Create the following Hospital Management System Class Hospital
Attributes: name, location, doctors (list), patients (list)
Methods:
1) add_doctor(doctor) : Adds a doctor to the hospital
2) add_patient(patient) : Adds a patient to the hospital
3) show_doctors() : Prints all doctors’ names and specialties
4) show_patients() : Prints all patients’ names and illnesses
 Class Doctor (inherits from Hospital)
Attributes: name, specialty, patients (list of Patient objects underthisdoctor)
Methods:
1) assign_patient(patient): Adds a patient to the doctor’s list
2) show_patients() : Shows all patients assigned to this doctor
 Class Patient (inherits from Hospital)
Attributes: name, age, illness, doctor (assigned doctor)
Methods:
1) assign_doctor(doctor) : Assigns a doctor to the patient
2) show_info() : Displays patient details including assigneddoctor

````python
books = []
borrowed_books = {} #Set to prevent duplicates

class Library():
    def __init__(self,name,books_list):
        self.name = name
        self.books = books_list
    def show_books(self):
        print(f"Available books in {self.name} are:")
        for book in self.books:
            print(book)
    def borrow_book(self):
        book = input("Enter the book name you want to borrow:")
        if book in self.books:
            self.books.remove(book)
            borrowed_books[book] = True
            print(f"{book} has been borrowed.")
        else:
            print(f"Sorry, {book} is not available in the library.")
    def return_book(self):
        book = input("Enter the book name you want to return:")
        if book in borrowed_books:
            del borrowed_books[book] #3l4an nsheel el book mn el borrowed books
            self.books.append(book)
            print(f"{book} has been returned.")
        else:
            print(f"Sorry, {book} was not borrowed from this library.")
    def show_borrowed_books(self):
        print(f"Borrowed books from {self.name} are:")
        for book in borrowed_books:
            print(book)
    def display(self):
        print(f"Library Name: {self.name}")
        print(f"Available Books: {self.books}")
        print(f"Borrowed Books: {list(borrowed_books.keys())}")
w = True
while(w):
    x = True
    print(f"Welcome to the library system!\nPlease enter the books or 'quit' to exit:")
    while(x):
        book = input("Enter the book name (or 'quit' to exit):")
        if book == 'quit':
            x = False
        else:
            books.append(book)
    depi = Library("DEPI Library",books)
    x = True
    while(x):
        print(f"========== Library Management System ==========\n"
                "1. Show Available Books\n"
                "2. Borrow Book\n"
                "3. Return Book\n"
                "4. Show Borrowed Books\n"
                "5. Exit\n")
        choice = int(input("Please enter your choice:"))
        if choice == 1:
            depi.show_books()
        elif choice == 2:
            depi.borrow_book()
        elif choice == 3:
            depi.return_book()
        elif choice == 4:
            depi.show_borrowed_books()
        elif choice == 5:
            print(f"Thank you for using the library system!")
            w = False
        else:
            print(f"Invalid choice! Please try again.")
        
````

## 2) Create the following hospital management system
- Class Hospital
- Class Doctor
- Class Patient

````python
class Hospital:
    def __init__(self, name, location):
        self.name = name
        self.location = location
        self.doctors = []
        self.patients = []
    def add_doctor(self, doctor):
        if doctor not in self.doctors:
            self.doctors.append(doctor)
            print(f"Dr. {doctor.name} has been added.")
        else:
            print(f"Sorry, Dr. {doctor.name} is already available in the hospital.")
    def add_patient(self, patient):
        if patient not in self.patients:
            self.patients.append(patient)
            print(f"Patient {patient.name} has been added.")
        else:
            print(f"Sorry, {patient.name} is already available in the hospital.")
    def show_doctors(self):
        print(f"Available doctors in {self.name}:")
        for doctor in self.doctors:
            print(f"- Dr. {doctor.name} (Specialty: {doctor.specialty})")
    def show_patients(self):
        print(f"Available patients in {self.name}:")
        for patient in self.patients:
            print(f"- {patient.name} (Illness: {patient.illness})")
class Doctor(Hospital):
    def __init__(self, name, specialty):
        self.name = name
        self.specialty = specialty
        self.patients = []
    def assign_patient(self, patient):
        if patient not in self.patients:
            self.patients.append(patient)
            print(f"{patient.name} has been assigned to Dr. {self.name}.")
        else:
            print(f"Sorry, {patient.name} is already assigned to Dr. {self.name}.")
    def show_patients(self):
        print(f"Patients assigned to Dr. {self.name} are:")
        for patient in self.patients:
            print(f"- {patient.name}")
class Patient(Hospital):
    def __init__(self, name, age, illness, doctor=None):
        self.name = name
        self.age = age
        self.illness = illness
        self.doctor = doctor
    def assign_doctor(self, doctor):
        self.doctor = doctor
        print(f"Dr. {doctor.name} has been assigned to {self.name}.")
    def show_info(self):
        doc_name = self.doctor.name if self.doctor else "No doctor assigned"
        print(f"Patient Name: {self.name}\nAge: {self.age}\nIllness: {self.illness}\nDoctor: {doc_name}")
hospital = Hospital("DEPI Hospital", "Alexandria")
doc1 = Doctor("Smith", "Cardiology")
pat1 = Patient("John Doe", 45, "Heart Disease")
while True:
    print(f"\nChoose an option:\n1. Add Doctor\n2. Add Patient\n3. Show Doctors\n4. Show Patients\n5. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        hospital.add_doctor(doc1)
    elif choice == 2:
        hospital.add_patient(pat1)
        doc1.assign_patient(pat1)
        pat1.assign_doctor(doc1)
    elif choice == 3:
        hospital.show_doctors()
    elif choice == 4:
        hospital.show_patients()
    elif choice == 5:
        print("Exiting system...")
        break
    else:
        print("Invalid choice! Please try again.")
````
